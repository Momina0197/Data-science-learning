```python
df = pd.read_csv('netflix_titles.csv', chunksize=1000)

categories = ['type', 'country']
output = pd.DataFrame()

for chunk in df:
    details = chunk[categories].copy()
    details['count'] = 1
    summary = details.groupby(categories).sum().reset_index()
    output = pd.concat([output, summary], ignore_index=True)

final_output = output.groupby(categories)['count'].sum().reset_index()
final_output = final_output.sort_values('count', ascending=False)
display(final_output.head(10))
```

```python
final_output.to_csv('aggregated-shows-info.csv', index=False)
```
